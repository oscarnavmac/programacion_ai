## 2. Colecciones

Elegir una colección es una decisión de diseño:

| Tipo | Orden | Mutable | Uso habitual |
|---|---|---|---|
| `list` | Sí | Sí | Secuencia que cambia |
| `tuple` | Sí | No | Registro corto o valor compuesto |
| `dict` | Sí* | Sí | Asociación clave → valor |
| `set` | No significativo | Sí | Valores únicos y pertenencia rápida |

Nota: los diccionarios conservan el orden de inserción, pero se consultan por clave.

```python
etiquetas = ["positivo", "neutral", "negativo", "positivo"]
coordenada = (20.97, -89.62)
prediccion = {"texto": "Excelente servicio", "etiqueta": "positivo", "score": 0.97}
etiquetas_unicas = set(etiquetas)

print(etiquetas[0], etiquetas[-1], etiquetas[1:3])
print(coordenada)
print(prediccion["etiqueta"], prediccion.get("idioma", "desconocido"))
print(etiquetas_unicas)
```

### Mutabilidad, alias y copia

Dos nombres pueden apuntar a la misma lista. Si queremos una colección
independiente debemos copiarla. Este detalle explica muchos errores silenciosos
en pipelines de datos.

```python
original = ["dato-a", "dato-b"]
alias = original
copia = original.copy()

alias.append("dato-c")
copia.append("dato-independiente")

print("original:", original)
print("alias:   ", alias)
print("copia:   ", copia)
print("¿Mismo objeto?", original is alias, original is copia)
```

## 3. Decisiones y truthiness

En una condición, Python interpreta como falsos `False`, `None`, cero y las
colecciones vacías. Esto permite escribir condiciones directas, siempre que el
significado sea claro.

```python
respuesta_modelo = ""

if not respuesta_modelo:
    estado = "sin respuesta"
elif len(respuesta_modelo) < 20:
    estado = "respuesta breve"
else:
    estado = "respuesta disponible"

print(estado)
```

### `match` cuando comparamos formas conocidas

`match` está disponible desde Python 3.10. No sustituye a todos los `if`, pero
puede expresar con claridad varios casos discretos.

```python
estado_http = 429

match estado_http:
    case 200:
        mensaje = "respuesta correcta"
    case 401 | 403:
        mensaje = "problema de autorización"
    case 429:
        mensaje = "demasiadas solicitudes"
    case _:
        mensaje = "estado no contemplado"

print(mensaje)
```

## 4. Ciclos y comprehensions

Un `for` recorre elementos, no índices. Usa `enumerate` cuando también necesites
la posición y `zip` para avanzar por varias secuencias a la vez.

```python
textos = ["muy bueno", "regular", "excelente"]
scores = [0.82, 0.51, 0.96]

for posicion, (texto, score) in enumerate(zip(textos, scores), start=1):
    print(f"{posicion}. {texto:<12} → {score:.0%}")
```

Las comprehensions sirven para transformaciones pequeñas y legibles. Si la
lógica requiere varias decisiones o efectos secundarios, un ciclo explícito
suele ser mejor.

```python
scores_altos = [score for score in scores if score >= 0.80]
longitudes = {texto: len(texto) for texto in textos}
iniciales = {texto[0] for texto in textos}

print(scores_altos)
print(longitudes)
print(iniciales)
```

## 5. Funciones: nombrar una transformación

Una función bien enfocada reduce repetición y permite probar una idea de forma
aislada. Sus parámetros son entradas; `return` produce una salida. Una función
sin `return` explícito devuelve `None`.

```python
def normalizar_texto(texto, *, minusculas=True):
    '''Elimina espacios externos y opcionalmente convierte a minúsculas.'''
    resultado = texto.strip()
    if minusculas:
        resultado = resultado.lower()
    return resultado


print(normalizar_texto("  Hola, IA  "))
print(normalizar_texto("  Hola, IA  ", minusculas=False))
```

El `*` vuelve explícitos los argumentos posteriores: debemos escribir
`minusculas=False`. La próxima sesión añadirá anotaciones de tipo y Pydantic;
hoy nos concentramos en el comportamiento.

```python
def resumir_scores(valores):
    if not valores:
        return 0.0, 0.0, 0.0
    return min(valores), sum(valores) / len(valores), max(valores)


minimo, promedio, maximo = resumir_scores(scores)
print(f"mín={minimo:.2f} · promedio={promedio:.2f} · máx={maximo:.2f}")
```

## 7. Excepciones: leer antes de corregir

Cuando Python no puede continuar, muestra un **traceback**. Léelo desde la última
línea:

1. tipo de excepción (`ValueError`, `KeyError`, `TypeError`, etc.);
2. mensaje concreto;
3. línea de nuestro código donde ocurrió;
4. cadena de llamadas que llevó hasta allí.

Captura sólo excepciones que puedas manejar. Un `except:` vacío oculta errores
reales y dificulta el diagnóstico.

```python
def convertir_score(valor):
    try:
        score = float(valor)
    except (TypeError, ValueError) as error:
        print(f"No se pudo convertir {valor!r}: {type(error).__name__}: {error}")
        return None

    if not 0 <= score <= 1:
        raise ValueError(f"El score debe estar entre 0 y 1; se recibió {score}")
    return score


print(convertir_score("0.87"))
print(convertir_score("desconocido"))
```

## 8. Clases: sólo cuando existe un concepto con estado

Una clase combina datos y comportamiento. No es un requisito para escribir buen
Python: si una función resuelve el problema con claridad, úsala. Este ejemplo
representa una plantilla que conserva un formato y sabe renderizarlo.

```python
class PlantillaPrompt:
    def __init__(self, plantilla):
        self.plantilla = plantilla

    def renderizar(self, **valores):
        return self.plantilla.format(**valores)


clasificador = PlantillaPrompt(
    "Clasifica el texto como positivo, neutral o negativo: {texto}"
)
print(clasificador.renderizar(texto="El producto llegó a tiempo"))
```

## 9. Mini-caso: limpiar y resumir predicciones

Integraremos diccionarios, listas, control de flujo, funciones, excepciones y
`Counter`. Los datos imitan una respuesta externa: algunos valores están como
texto y una fila es inválida.

```python
predicciones_crudas = [
    {"texto": "Excelente", "etiqueta": "positivo", "score": "0.98"},
    {"texto": "No estuvo mal", "etiqueta": "neutral", "score": 0.66},
    {"texto": "Llegó roto", "etiqueta": "negativo", "score": "0.93"},
    {"texto": "Sin datos", "etiqueta": "desconocido", "score": None},
]


def limpiar_prediccion(registro):
    score = convertir_score(registro.get("score"))
    etiqueta = registro.get("etiqueta", "desconocido").strip().lower()
    if score is None or etiqueta not in {"positivo", "neutral", "negativo"}:
        return None
    return {
        "texto": registro.get("texto", "").strip(),
        "etiqueta": etiqueta,
        "score": score,
    }


predicciones = []
for registro in predicciones_crudas:
    if (limpio := limpiar_prediccion(registro)) is not None:
        predicciones.append(limpio)

conteo_etiquetas = Counter(item["etiqueta"] for item in predicciones)
score_promedio = mean(item["score"] for item in predicciones)

print("Registros válidos:", len(predicciones))
print("Etiquetas:", conteo_etiquetas)
print(f"Confianza promedio: {score_promedio:.1%}")
```
