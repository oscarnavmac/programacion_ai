````python
original = ["dato-a", "dato-b"]
alias = original
copia = original.copy()

alias.append("dato-c")
copia.append("dato-independiente")

print("original:", original)
print("alias:   ", alias)
print("copia:   ", copia)
print("¿Mismo objeto?", original is alias, original is copia)
````

## 4. Ciclos y comprehensions

Un `for` recorre los elementos de un iterable. En una lista puedes recorrer los
valores directamente; no hace falta construir sus índices. Usa `enumerate` cuando
también necesites la posición y `zip` para avanzar por varias secuencias a la vez.
Por defecto, `zip` se detiene al agotarse la secuencia más corta. Si necesitas
comprobar que tengan la misma longitud, usa `zip(..., strict=True)`.

````python
textos = ["muy bueno", "regular", "excelente"]
scores = [0.82, 0.51, 0.96]

for posicion, (texto, score) in enumerate(zip(textos, scores), start=1):
    print(f"{posicion}. {texto:<12} → {score:.0%}")
````

Las comprehensions sirven para transformaciones pequeñas y legibles. Si la
lógica requiere varias decisiones o efectos secundarios, un ciclo explícito
suele ser mejor.

````python
scores_altos = [score for score in scores if score >= 0.80]
longitudes = {texto: len(texto) for texto in textos}
iniciales = {texto[0] for texto in textos}

print(scores_altos)
print(longitudes)
print(iniciales)
````

## 5. Funciones: nombrar una transformación

Una función bien enfocada reduce repetición y permite probar una idea de forma
aislada. Sus parámetros son entradas; `return` produce una salida. Una función
sin `return` explícito devuelve `None`.

````python
def normalizar_texto(texto, *, minusculas=True):
    '''Elimina espacios externos y opcionalmente convierte a minúsculas.'''
    resultado = texto.strip()
    if minusculas:
        resultado = resultado.lower()
    return resultado


print(normalizar_texto("  Hola, IA  "))
print(normalizar_texto("  Hola, IA  ", minusculas=False))
````

El `*` vuelve explícitos los argumentos posteriores: debemos escribir

### Ordenar usando una función

`sorted` crea una lista nueva; `.sort()` modifica la lista existente. El argumento
`key` recibe una función que calcula el criterio de orden. Pasamos la función
sin llamarla: Python la aplicará a cada elemento.

````python
evaluaciones = [
    {"modelo": "a", "score": 0.7},
    {"modelo": "b", "score": 0.95},
    {"modelo": "c", "score": 0.8},
]

def obtener_score(evaluacion):
    return evaluacion["score"]

ranking = sorted(evaluaciones, key=obtener_score, reverse=True)
print(ranking)
# lambda permite expresar una función pequeña en el mismo lugar.
assert ranking == sorted(evaluaciones, key=lambda e: e["score"], reverse=True)
````

### Implementación de apoyo para continuar

Después de intentar el ejercicio, puedes ejecutar esta implementación para seguir
con el taller. `raise` señala un error y `isinstance` comprueba si un objeto
pertenece a un tipo. El encadenamiento `from error` conserva la causa original.

Esta función permanece como código ejecutable porque los ejemplos posteriores
la necesitan. La solución con sus comprobaciones está en Markdown en el anexo.

````python
def validar_score(valor):
    if isinstance(valor, bool):
        raise ValueError("Un booleano no es un score")
    if not isinstance(valor, (int, float, str)):
        raise ValueError("El score debe ser int, float o str")
    try:
        score = float(valor)
    except (ValueError, OverflowError) as error:
        raise ValueError("Score no convertible a número") from error
    if not 0 <= score <= 1:
        raise ValueError("Score fuera del intervalo [0, 1]")
    return score
````

permanece ejecutable. La solución del ejercicio se consulta en el anexo.

````python
def limpiar_prediccion(registro):
    if not isinstance(registro, dict):
        raise ValueError("El registro debe ser un diccionario")
    texto = registro.get("texto")
    etiqueta = registro.get("etiqueta")
    if not isinstance(texto, str):
        raise ValueError("El texto debe ser una cadena")
    if not isinstance(etiqueta, str):
        raise ValueError("La etiqueta debe ser una cadena")
    etiqueta = etiqueta.strip().lower()
    if etiqueta not in {"positivo", "neutral", "negativo"}:
        raise ValueError("Etiqueta no reconocida")
    return {
        "texto": texto.strip(),
        "etiqueta": etiqueta,
        "score": validar_score(registro.get("score")),
    }
````

## 9. POO: objetos que conservan estado

Una **clase** define cómo construir objetos y qué operaciones ofrecen. Cada
**instancia** es un objeto concreto. Sus **atributos** guardan datos y sus
**métodos** son funciones asociadas al objeto.

`__init__` inicializa una instancia; `self` representa esa instancia dentro de los
métodos. Al llamar `contador.agregar()`, Python pasa `contador` como `self`.

````python
class Contador:
    def __init__(self, nombre):
        self.nombre = nombre
        self.total = 0

    def agregar(self):
        self.total += 1

    def reiniciar(self):
        self.total = 0

contador_a = Contador("lote A")
contador_b = Contador("lote B")
contador_a.agregar()
contador_a.agregar()
print(contador_a.nombre, contador_a.total)
print(contador_b.nombre, contador_b.total)
assert contador_b.total == 0
````

Contrato de `agregar`: devuelve `True` si acepta el registro y `False` si lo rechaza.
En un rechazo guarda la posición original y el motivo. `resumen` devuelve cantidad,
conteos y promedio; el promedio es `None` cuando no hay registros válidos.

El prefijo `_` indica atributos de uso interno por convención; no impide acceder
a ellos. Usa los métodos para mantener las reglas de la clase.

````python
predicciones_crudas = [
    {"texto": " Excelente ", "etiqueta": " POSITIVO ", "score": "0.98"},
    {"texto": "Regular", "etiqueta": "neutral", "score": 0.66},
    {"texto": "Llegó roto", "etiqueta": "negativo", "score": "0.93"},
    {"texto": "Sin score", "etiqueta": "neutral", "score": None},
    {"texto": "Fuera de rango", "etiqueta": "positivo", "score": 1.2},
    {"texto": "Sin etiqueta", "etiqueta": None, "score": 0.8},
    {"texto": "Cero válido", "etiqueta": "neutral", "score": 0.0},
    {"texto": "Falta clave", "etiqueta": "positivo"},
]
````

````python
class RegistroPredicciones:
    def __init__(self, nombre):
        self.nombre = nombre
        self._validas = []
        self._rechazos = []

    def agregar(self, registro, posicion):
        try:
            limpio = limpiar_prediccion(registro)
        except ValueError as error:
            self._rechazos.append({"posicion": posicion, "motivo": str(error)})
            return False
        limpio["posicion"] = posicion
        self._validas.append(limpio)
        return True

    def resumen(self):
        scores = [registro["score"] for registro in self._validas]
        return {
            "validas": len(self._validas),
            "rechazadas": len(self._rechazos),
            "conteos": dict(Counter(r["etiqueta"] for r in self._validas)),
            "promedio": sum(scores) / len(scores) if scores else None,
        }

    def rechazos(self):
        return [rechazo.copy() for rechazo in self._rechazos]

lote = RegistroPredicciones("Demostración")
for posicion, registro in enumerate(predicciones_crudas):
    aceptada = lote.agregar(registro, posicion)
    print(posicion, "aceptada" if aceptada else "rechazada")
print(lote.resumen())
print(lote.rechazos())
````

def validar_score(valor):
    if isinstance(valor, bool):
        raise ValueError("Un booleano no es un score")
    if not isinstance(valor, (int, float, str)):
